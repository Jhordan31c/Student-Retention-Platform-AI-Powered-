```python
# Cargar dataset
df = pd.read_csv('../data/raw/student_data.csv')

print(f"📊 Dataset cargado: {df.shape[0]} filas × {df.shape[1]} columnas")
print(f"\n📋 Columnas:")
for i, col in enumerate(df.columns, 1):
    print(f"  {i:2d}. {col}")
```

## 🎯 4. ANÁLISIS DEL TARGET (Variable Objetivo)

```python
# Identificar columna target
target_col = 'Target'  # Ajustar si el nombre es diferente

if target_col in df.columns:
    print(f"✅ Target encontrado: {target_col}")
    print(f"\n📊 Distribución de clases:")
    print(df[target_col].value_counts())
    print(f"\n📈 Porcentajes:")
    print(df[target_col].value_counts(normalize=True).mul(100).round(2))
else:
    print("⚠️ No se encontró la columna 'Target'. Columnas disponibles:")
    print(df.columns.tolist())
```

```python
# Visualización interactiva del target
if target_col in df.columns:
    fig = px.pie(
        df[target_col].value_counts().reset_index(),
        values='count',
        names=target_col,
        title='📊 Distribución de Clases (Target)',
        hole=0.4,
        color_discrete_sequence=px.colors.qualitative.Set3
    )
    fig.update_traces(textposition='inside', textinfo='percent+label')
    fig.update_layout(height=500)
    
    fig.show()
    
    # Análisis de desbalanceo
    class_counts = df[target_col].value_counts()
    majority_class = class_counts.max()
    minority_class = class_counts.min()
    imbalance_ratio = majority_class / minority_class
    
    print(f"\n⚖️ ANÁLISIS DE DESBALANCEO:")
    print(f"  • Clase mayoritaria: {class_counts.idxmax()} ({majority_class} muestras)")
    print(f"  • Clase minoritaria: {class_counts.idxmin()} ({minority_class} muestras)")
    print(f"  • Ratio de desbalanceo: {imbalance_ratio:.2f}:1")
    
    if imbalance_ratio > 3:
        print(f"  ⚠️ DATASET DESBALANCEADO - Considerar SMOTE o class_weight")
    else:
        print(f"  ✅ Dataset relativamente balanceado")
```

## 🔎 5. DETECCIÓN DE VALORES FALTANTES

```python
# Análisis de missing values
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({
    'Missing_Count': missing,
    'Missing_Percentage': missing_pct
}).sort_values('Missing_Count', ascending=False)

missing_df = missing_df[missing_df['Missing_Count'] > 0]

if len(missing_df) > 0:
    print("⚠️ COLUMNAS CON VALORES FALTANTES:")
    print(missing_df)
    
    # Visualización
    fig = px.bar(
        missing_df.reset_index(),
        x='index',
        y='Missing_Percentage',
        title='📉 Porcentaje de Valores Faltantes por Columna',
        labels={'index': 'Columna', 'Missing_Percentage': '% Faltante'}
    )
    fig.update_layout(xaxis_tickangle=-45, height=500)
    fig.show()
else:
    print("✅ NO HAY VALORES FALTANTES - Dataset completo")
```

## 📊 6. ANÁLISIS DE FEATURES NUMÉRICAS

```python
# Identificar columnas numéricas
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
print(f"📊 Features numéricas: {len(numeric_cols)}")
print(numeric_cols[:10], "..." if len(numeric_cols) > 10 else "")
```

```python
# Distribuciones de features numéricas clave (primeras 6)
key_numeric_features = numeric_cols[:6]

fig = make_subplots(
    rows=2, cols=3,
    subplot_titles=key_numeric_features,
    vertical_spacing=0.12,
    horizontal_spacing=0.1
)

for i, col in enumerate(key_numeric_features, 1):
    row = (i - 1) // 3 + 1
    col_pos = (i - 1) % 3 + 1
    
    fig.add_trace(
        go.Histogram(x=df[col], name=col, showlegend=False),
        row=row, col=col_pos
    )

fig.update_layout(height=600, title_text="📊 Distribuciones de Features Numéricas Clave")
fig.show()
```

## 🎨 7. ANÁLISIS DE CORRELACIONES

```python
# Matriz de correlación
correlation_matrix = df[numeric_cols].corr()

# Heatmap interactivo
fig = px.imshow(
    correlation_matrix,
    text_auto='.2f',
    aspect='auto',
    color_continuous_scale='RdBu_r',
    color_continuous_midpoint=0,
    title='🔥 Matriz de Correlación - Features Numéricas'
)
fig.update_layout(height=800, width=900)
fig.show()
```

```python
# Análisis de features numéricas vs target categórico
from scipy.stats import f_oneway

print("=" * 60)
print("ANÁLISIS: Features Numéricas vs Target Categórico")
print("=" * 60)

# Calcular ANOVA F-statistic para cada feature numérica
anova_results = []

for col in numeric_cols:
    if col != target_col:  # Excluir el target si estuviera como numérico
        groups = [df[df[target_col] == category][col].dropna() 
                  for category in df[target_col].unique()]
        
        # Solo calcular si hay suficientes datos
        if all(len(g) > 0 for g in groups):
            f_stat, p_value = f_oneway(*groups)
            anova_results.append({
                'Feature': col,
                'F_Statistic': f_stat,
                'P_Value': p_value,
                'Significativo': 'Sí' if p_value < 0.05 else 'No'
            })

# Crear DataFrame y ordenar por F-statistic
anova_df = pd.DataFrame(anova_results).sort_values('F_Statistic', ascending=False)

print("\nTop 15 Features más relacionadas con el Target (por ANOVA):")
print(anova_df.head(15).to_string(index=False))

# Visualización
top_15 = anova_df.head(15)
fig = px.bar(
    top_15,
    x='F_Statistic',
    y='Feature',
    orientation='h',
    title='Top 15 Features por Importancia (ANOVA F-Statistic)',
    labels={'F_Statistic': 'F-Statistic', 'Feature': 'Feature'},
    color='F_Statistic',
    color_continuous_scale='Viridis'
)
fig.update_layout(height=600, yaxis={'categoryorder':'total ascending'})
fig.show()

print(f"\nInterpretación:")
print(f"- F-Statistic alto = La feature separa bien las clases del target")
print(f"- P-Value < 0.05 = Relación estadísticamente significativa")
print(f"\nFeatures significativas: {anova_df[anova_df['P_Value'] < 0.05].shape[0]} de {len(anova_df)}")
```

## 📦 8. DETECCIÓN DE OUTLIERS

```python
# Boxplots para detectar outliers (primeros 6 features)
key_features_for_outliers = numeric_cols[:6]

fig = make_subplots(
    rows=2, cols=3,
    subplot_titles=key_features_for_outliers,
    vertical_spacing=0.15,
    horizontal_spacing=0.1
)

for i, col in enumerate(key_features_for_outliers, 1):
    row = (i - 1) // 3 + 1
    col_pos = (i - 1) % 3 + 1
    
    fig.add_trace(
        go.Box(y=df[col], name=col, showlegend=False),
        row=row, col=col_pos
    )

fig.update_layout(height=600, title_text="📦 Boxplots - Detección de Outliers")
fig.show()
```

```python
# Separar features numéricas continuas de categóricas codificadas
continuous_features = [
    'Age at enrollment',
    'Previous qualification (grade)',
    'Admission grade',
    'Curricular units 1st sem (credited)',
    'Curricular units 1st sem (enrolled)',
    'Curricular units 1st sem (evaluations)',
    'Curricular units 1st sem (approved)',
    'Curricular units 1st sem (grade)',
    'Curricular units 2nd sem (credited)',
    'Curricular units 2nd sem (enrolled)',
    'Curricular units 2nd sem (evaluations)',
    'Curricular units 2nd sem (approved)',
    'Curricular units 2nd sem (grade)',
    'Unemployment rate',
    'Inflation rate',
    'GDP'
]

# Filtrar solo las que existen en tu dataset
continuous_features = [col for col in continuous_features if col in df.columns]

print(f"Features numéricas continuas: {len(continuous_features)}")
print(continuous_features)
```

```python
# Análisis de outliers SOLO en variables continuas
print("ANÁLISIS DE OUTLIERS (Solo variables continuas):")
print("="*60)
outlier_summary = []

for col in continuous_features:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    count = len(outliers)
    pct = (count / len(df) * 100)
    
    outlier_summary.append({
        'Feature': col,
        'Outliers_Count': count,
        'Outliers_Pct': round(pct, 2)
    })
    
    if pct > 5:
        print(f"⚠️ {col}: {count} outliers ({pct:.2f}%)")

outlier_df = pd.DataFrame(outlier_summary).sort_values('Outliers_Pct', ascending=False)
print(f"\nResumen:")
print(outlier_df)
```

```python
# Buscar valores imposibles en calificaciones (deben estar 0-20)
print("Verificación de rangos válidos:")
print(f"Grades 1st sem: {df['Curricular units 1st sem (grade)'].min()} - {df['Curricular units 1st sem (grade)'].max()}")
print(f"Grades 2nd sem: {df['Curricular units 2nd sem (grade)'].min()} - {df['Curricular units 2nd sem (grade)'].max()}")
print(f"Age: {df['Age at enrollment'].min()} - {df['Age at enrollment'].max()}")
```
